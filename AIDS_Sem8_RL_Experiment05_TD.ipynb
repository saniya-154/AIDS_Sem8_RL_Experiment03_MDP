```python
print("Saniya Sayyed | 221A014 | 37")
import numpy as np
import random

#Gridworld Envirnment
class GridWorld:
    def __init__(self, size=4, start=(0,0), goal=(3,3)):
        self.size = size
        self.start = start
        self.goal = goal
        self.state = start
        self.actions = [(0,1),(1,0),(0,-1),(-1,0)] # Right, Up, Left, Down

    def reset(self):
        self.state = self.start
        return self.state

    def step(self, action):
        next_state = (self.state[0] + self.actions[action][0], self.state[1] + self.actions[action][1])
        if 0 <= next_state[0] < self.size and 0 <= next_state[1] < self.size:
          self.state = next_state
        reward = 1 if self.state == self.goal else -0.1
        done = self.state == self.goal
        return self.state, reward, done
```

```python
print("Saniya Sayyed | 221A014 | 37")
def monte_carlo_control(env, episodes=5000, gamma=0.9, epsilon=0.1):
    Q = {((i, j), a): 0 for i in range(env.size) for j in range(env.size) for a in range(4)}
    returns = {((i, j), a): [] for i in range(env.size) for j in range(env.size) for a in range(4)}

    for _ in range(episodes):
        state = env.reset()
        episode = []

        while True:
            action = random. choice(range(4)) if random.random() < epsilon else max(range(4), key=lambda a: Q[(state, a)])
            next_state, reward, done = env.step(action)
            episode.append((state, action, reward))
            state = next_state
            if done:
              break

        G = 0
        for state, action, reward in reversed(episode):
            G = gamma * G + reward
            if (state, action) not in [(s, a) for s, a, _ in episode[:-1]]:
              returns[(state, action)].append(G)
              Q[state, action] = np.mean(returns[(state, action)])

    return Q
```

### SARSA (State-Action-Reward-State-Action)

```python
# Temporal Differencing learning (SARSA)
print("Saniya Sayyed | 221A014 | 37")
def sarsa(env, episodes=5000, alpha=0.1, gamma=0.9, epsilon=0.1):
    Q = {((i, j), a): 0 for i in range(env.size) for j in range(env.size) for a in range(4)}

    for _ in range(episodes):
        state = env.reset()
        action = random.choice(range(4)) if random.random() < epsilon else max(range(4), key=lambda a: Q[(state, a)])

        while True:
            next_state, reward, done = env.step(action)
            next_action = random.choice(range(4)) if random.random() < epsilon else max(range(4), key=lambda a: Q[(next_state, a)])

            Q[(state, action)] += alpha * (reward + gamma * Q[(next_state, next_action)] - Q[(state, action)])
            state, action = next_state, next_action
            if done:
               break

    return Q
```

### Q-Learning Algorithm

```python
print("Saniya Sayyed | 221A014 | 37")
def q_learning(env, episodes=5000, alpha=0.1, gamma=0.9, epsilon=0.1):
  Q = {((i, j), a): 0 for i in range(env.size) for j in range(env.size) for a in range(4)}

  for _ in range(episodes):
    state = env.reset()

    while True:
      action = random.choice(range(4)) if random.random() < epsilon else max(range(4), key=lambda a: Q[(state, a)])
      next_state, reward, done = env.step(action)

      Q[(state, action)] += alpha * (reward + gamma * max(Q[(next_state, a)] for a in range(4)) - Q[(state, action)])
      state = next_state
      if done:
          break

    return Q
```

### Running the Training

```python
print("Saniya Sayyed | 221A014 | 37")
#Initialize Environment
env = GridWorld()

# Train with Monte Carlo
Q_mc = monte_carlo_control(env)

# Train with SARSA
Q_sarsa = sarsa(env)

# Train with Q-Learning
Q_qlearning = q_learning(env)
```

### Analysis of Q-values

*   **Monte Carlo Control**: In its current implementation, the Monte Carlo method did not learn effective Q-values within the initial samples shown, as many remain at 0. This is likely due to its episodic nature, requiring full episodes to complete for updates.
*   **SARSA**: This temporal difference control method shows more developed Q-values, suggesting continuous learning throughout episodes. The sample values indicate active learning.
*   **Q-Learning**: This also demonstrates some initial learning, though similar to Monte Carlo, some values are still at their initialized state. This could be due to the specific exploration strategy or the limited number of episodes shown.
*   **Update Mechanisms**: Monte Carlo updates its Q-values only after an entire episode is completed. In contrast, SARSA and Q-Learning are temporal difference methods that update at each step, potentially leading to faster learning in some environments by leveraging immediate rewards and future estimated rewards.
